# Local-compatible pretrained V3 / 个人电脑兼容的预训练V3
Portable setup is the default; A100 is not required. No training or private Drive mount.
默认普通运行模式，不要求A100，不训练，也不挂载作者的Drive。严格复现才使用 --runtime reference。

```
from pathlib import Path
import sys, subprocess, urllib.request, zipfile
DEST=Path('/content/ecc_wouaf_public_product')
DEST.mkdir(exist_ok=False)
archive=DEST/'source.zip'
urllib.request.urlretrieve('https://github.com/Montos-Qu/ecc-protected-wouaf/archive/refs/heads/main.zip', archive)
with zipfile.ZipFile(archive) as z:
    for name in z.namelist():
        assert (DEST/name).resolve().is_relative_to(DEST.resolve())
    z.extractall(DEST)
REPO=DEST/'ecc-protected-wouaf-main'
def product(*args):
    return subprocess.run([sys.executable,str(REPO/'wouaf.py'),*map(str,args)],cwd=REPO,check=True)
```

```
product('setup')  # Restore trained weights, runtime and demo assets; never train.
```

```
VERSION='v3'  # Final product by default; optional controls: 'v1' or 'v2'.
product('versions')
product('predict','--image',REPO/'demo/images/png8_control/positive_k11/10000000.png','--version',VERSION,'--output',REPO/'outputs/example_prediction.json')
```

## 可选：完整独立图像队列

只有需要从全部真实图片重新评估时才启用下一格的开关；默认保持False。环境和模型会复用，结果单独存放。
